# Customer Intelligence, CLV & Churn Prediction Platform

This notebook provides a compact walkthrough of the validated analytical outputs. The full reproducible pipeline lives in `Python/src/customer_intelligence`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == "Notebooks" else Path.cwd()
DATA = ROOT / "Data" / "Processed"
plt.style.use("seaborn-v0_8-whitegrid")
```

## Executive KPIs

```python
executive = pd.read_csv(DATA / "executive_kpis.csv")
executive
```

## RFM Portfolio

```python
segments = pd.read_csv(DATA / "segment_summary.csv").sort_values("predicted_clv_12m")
ax = segments.plot.barh(
    x="rfm_segment",
    y="predicted_clv_12m",
    color="#987CF5",
    figsize=(10, 5),
    legend=False,
)
ax.set_title("Predicted 12-Month CLV by RFM Segment")
ax.set_xlabel("TRY")
ax.set_ylabel("")
plt.tight_layout()
```

## Model Comparison

```python
churn_models = pd.read_csv(DATA / "churn_model_comparison.csv")
clv_models = pd.read_csv(DATA / "clv_model_comparison.csv")
display(churn_models[["model", "roc_auc", "pr_auc", "brier_score", "lift_at_10pct"]])
display(clv_models[["model", "mae", "rmse", "r2", "spearman_correlation"]])
```

## SHAP Drivers

```python
shap_global = pd.read_csv(DATA / "shap_global_importance.csv").head(10).sort_values("importance_share")
ax = shap_global.plot.barh(
    x="feature",
    y="importance_share",
    color="#F2556D",
    figsize=(10, 5),
    legend=False,
)
ax.set_title("Top Global Churn Drivers")
ax.set_xlabel("Importance Share")
ax.set_ylabel("")
plt.tight_layout()
```
